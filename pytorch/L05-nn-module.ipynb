#### **The NN Module**
The torch.nn module in PyTorch is a core library that provides a wide array of classes and functions designed to help developers build neural networks
efficiently and effectively. It abstracts the complexity of creating and training neural networks by offering pre-built layers, loss functions,
activation functions and other utilities, enabling us to focus on designing and experimenting with model architectures. 

#### **Key Components of torch.nn**
1. **Modules (Layers):**
    - **nn.Module:** The base class for all neural network modules. Our custom models and layers hodul be subclass of this class
    - **Common Layers:** Includes layers like nn.Linear (fully connected layer), nn.Conv2d (convolutional layer), nn.LSTM (recurrent layer), and many others
2. **Activation Functions:**
    - Funcitons like nn.ReLU, nn.Sigmoid, nn.Tanh etc. introduce non-linearies to the model, allowing it to learn complex patterns
3. **Loss Functions:**
    - Provides loss functions such as nn.CrossEntropyLoss, nn.MSELoss, and nn.NLLLoss to quantify the difference between the model's predictions and the actual targets
4. **Conatiner Modules:**
    - **nn.Sequential**: A sequential container ro stack layers in order.
5. **Regularization and Dropout:**
    - Layers like nn.Dropout and nn.BatchNorm2d help prevent overfitting and improve the model's ability to generate new data

#### Code Examples:

1. Binary classification problem
2. One neuron, five features, one output

```python
# create model class
import torch
import torch.nn as nn 

class Model(nn.Module): ## nn.Module is the base class, we need to inherit from this class for creating our model
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features, 1) # nn.Linear is used for the linear weight matrix [simply for wx + b], accepts parameters: (num_of_features, num_of_output)
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, features):
        out = self.linear(features) # we are calculating z = wx + b here
        out = self.sigmoid(out)
        
        return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = Model(features.shape[1])

# forward pass
model(features) # nn.Module class is configured as when we well call the instance, it will automatically call the forward() method
```

```python
print(model.linear.weight)
print(model.linear.bias)
```

```python
from torchinfo import summary
summary(model, input_size=(10, 5))
```

#### Neural network with hidden layers
- 5 features, 1 hidden layer with 3 neurons, one output
- ReLU in hidden layers

```python
# Create the complex class

class ModelWithLayers(nn.Module):
    def __init__(self, num_of_features):
        super().__init__()
        self.layer1 = nn.Linear(num_of_features, 3) # 3 is the number of output, which will be the hidden layer
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(3, 1) # from 3 neuron to 1 final output
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, features):
        out = self.layer1(features)
        out = self.relu(out)
        out = self.layer2(out)
        out = self.sigmoid(out)
        
        return out
```

```python
# create dataset
feats = torch.rand(10, 5)

# create a new model instance
model2 = ModelWithLayers(feats.shape[1])

# run forward pass
model2(feats)
```

```python
print(model2.layer2.weight)
print(model2.layer2.bias)
```

```python
from torchinfo import summary
summary(model2, input_size=(10, 5))
```

#### Sequential container
We can see, if the number of layer increases, it becomes cumbersome to create the layers and chaining them to the forward pass. Sequential container comes here to salvage us. 

```python
# Sequential Container
class SeqConModel(nn.Module):
    def __init__(self, num_of_features):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(num_of_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )
        
    def forward(self, features):
        out = self.network(features)
        
        return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = SeqConModel(features.shape[1])

# forward pass
model(features)
```
