```python
# this will return false in apple silicon devices, as CUDA is explicitly for NVIDA GPUs
if torch.cuda.is_available():
  print("GPU is available!")
  print(f"Using GPU: {torch.cuda.get_device_name(0)}")
else:
  print("GPU is not available. Using CPU")
```

```python
# For modern apple silicon chips
if torch.backends.mps.is_available():
    device = torch.device("mps") # we will need this device to force tensors to use mps
    print("Using apple silicon GPU acceleration (MPS)")
else:
    device = torch.device("cpu")
    print("MPS device not found. Using CPU")
```

# Creating a Tensor

```python
# using empty
## it creates empty tensor, (2, 3) is the shape. 
## empty function allocates memory for us, as the given shape (dimension)
## it will not necessarily have all 0s, it will allocate the existing values
## of the allocated space
a = torch.empty(2, 3) 
```
