# Creating a Tensor

```python
# using empty
## it creates empty tensor, (2, 3) is the shape. 
## empty function allocates memory for us, as the given shape (dimension)
## it will not necessarily have all 0s, it will allocate the existing values
## of the allocated space
a = torch.empty(2, 3) 
```

```python
# using type
type(a)
```

```python
# using zeros
## it will give a tensor with all 0 values of a given shape
torch.zeros(2, 3)
```

```python
# using ones
torch.ones(2, 3)
```

```python
# using rand
torch.rand(2, 3)
```

```python
# using seed
## this will retain the random tensor over the same ipynb file, for a same seed number, in this case the number is 100
torch.manual_seed(100)
torch.rand(2, 3)
```

```python
# verifying the seed
torch.manual_seed(100)
torch.rand(2, 3)
```

```python
# for creating custom tensors
torch.tensor([[1, 2, 3], [4, 5, 6]])
```

```python
# other ways
# arange -> creates tensors of a range, torch.arange(start, end, diff)
torch.arange(0, 10, 2)

# using linspace -> evenly spaced values: torch.linspace(start, end, even_space_between_each)
torch.linspace(0, 10, 10)

# usingb eye -> eye for identity matrix, size should be given to the parameter
torch.eye(5)

# torch.full -> fill up a full matrix with a same value
torch.full((3, 3), 5)
```

## Mathematical Operations

### Scaler Operations

```python
x = torch.rand(2, 2)
x
```

```python
# addition
print(f"Add 2 to each cell: {x + 2}")

# subtraction
print(f"Subtract 2 from each cell: {x - 2}")


# multiplication
print(f"Multiply 2 with each cell: {x * 2}")

# division
print(f"Divide 2 from each cell: {x / 2}")

# int division
print(f"Integer division: {(x * 100) // 3}")

# mod
print(f"Mod 2: {((x * 100) // 3) % 2}")

# power
print(f"Square: {x**2}")
```

### Element wise operation

```python
a = torch.rand(2, 3)
b = torch.rand(2, 3)

print(a)
print(b)
```

```python
# add
print(f"Adding two tensors: {a + b}")

# sub
print(f"Subtracting: {a - b}")

# multiplication
print(f"Multiplication: {a * b}")

# division
print(f"Division: {a / b}")

# power
print(f"Power: {a**b}")

# mod
print(f"Mod: {a % b}")
```

```python
c = torch.tensor([2, 3, 4, 5])

# to make tensor in a certain range (min, max). The elements greater than max will become max, the elements less than min will become min
torch.clamp(c, min=3, max=4)
```

### Reduction Operation

```python
e = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
e
```

```python
# sum of all of the elements of tensor e
print(f"Sum of all the elements: {torch.sum(e)}")

# sum along columns
print(f"Sum of each column: {torch.sum(e, dim=0)}") # dim 0 will make a tensor of 1 dimension of each columns sum

# sum along rows
print(f"Sum along each row: {torch.sum(e, dim=1)}")
```

```python
# mean -> only works when datatype is either a floating point or complex dtype
print(f"Mean of all elements: {torch.mean(e)}")

# mean along col
print(f"Mean of each column: {torch.mean(e, dim=0)}")

# mean along row
print(f"Mean of each row: {torch.mean(e, dim=1)}")
```

```python
# median
torch.median(e)
```

```python
# argmax -> gets the index/position of the maximum item
torch.argmax(e)
```

### Matrix Operation

```python
f = torch.randint(size=(2, 3), low=0, high=10)
g = torch.randint(size=(3, 2), low=0, high=10)

print(f)
print(g)
```

```python
torch.matmul(f, g)
```

```python
# dot product
v1 = torch.tensor([1, 2])
v2 = torch.tensor([3, 4])

torch.dot(v1, v2)
```

```python
# transpose
print(f"Swap columns with row: {torch.transpose(f, 0, 1)}") # 0, 1 means swap column with row
```

### Comparison Operations

```python
i = torch.randint(size=(2, 3), low=0, high=10)
j = torch.randint(size=(2, 3), low=0, high=10)

print(i)
print(j)
```

```python
# greater than
print(f"greater than operation: {i > j}")

# less than
print(f"less than operation: {i < j}")

# equal to
print(f"equal to operation: {i == j}")

# not equal to
print(f"Not Equal to operation: {i != j}")
```
