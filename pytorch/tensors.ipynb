## Mathematical Operations

### Scaler Operations

````python
x = torch.rand(2, 2)
x
````

````python
# addition
print(f"Add 2 to each cell: {x + 2}")

# subtraction
print(f"Subtract 2 from each cell: {x - 2}")


# multiplication
print(f"Multiply 2 with each cell: {x * 2}")

# division
print(f"Divide 2 from each cell: {x / 2}")

# int division
print(f"Integer division: {(x * 100) // 3}")

# mod
print(f"Mod 2: {((x * 100) // 3) % 2}")

# power
print(f"Square: {x**2}")
````

### Element wise operation

````python
a = torch.rand(2, 3)
b = torch.rand(2, 3)

print(a)
print(b)
````

````python
# add
print(f"Adding two tensors: {a + b}")

# sub
print(f"Subtracting: {a - b}")

# multiplication
print(f"Multiplication: {a * b}")

# division
print(f"Division: {a / b}")

# power
print(f"Power: {a**b}")

# mod
print(f"Mod: {a % b}")
````

````python
c = torch.tensor([2, 3, 4, 5])

# to make tensor in a certain range (min, max). The elements greater than max will become max, the elements less than min will become min
torch.clamp(c, min=3, max=4)
````

### Reduction Operation

````python
e = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
e
````

````python
# sum of all of the elements of tensor e
print(f"Sum of all the elements: {torch.sum(e)}")

# sum along columns
print(f"Sum of each column: {torch.sum(e, dim=0)}") # dim 0 will make a tensor of 1 dimension of each columns sum

# sum along rows
print(f"Sum along each row: {torch.sum(e, dim=1)}")
````

````python
# mean -> only works when datatype is either a floating point or complex dtype
print(f"Mean of all elements: {torch.mean(e)}")

# mean along col
print(f"Mean of each column: {torch.mean(e, dim=0)}")

# mean along row
print(f"Mean of each row: {torch.mean(e, dim=1)}")
````

````python
# median
torch.median(e)
````

````python
# argmax -> gets the index/position of the maximum item
torch.argmax(e)
````

### Matrix Operation

````python
f = torch.randint(size=(2, 3), low=0, high=10)
g = torch.randint(size=(3, 2), low=0, high=10)

print(f)
print(g)
````

````python
torch.matmul(f, g)
````

````python
# dot product
v1 = torch.tensor([1, 2])
v2 = torch.tensor([3, 4])

torch.dot(v1, v2)
````

````python
# transpose
print(f"Swap columns with row: {torch.transpose(f, 0, 1)}") # 0, 1 means swap column with row
````

### Comparison Operations

````python
i = torch.randint(size=(2, 3), low=0, high=10)
j = torch.randint(size=(2, 3), low=0, high=10)

print(i)
print(j)
````

````python
# greater than
print(f"greater than operation: {i > j}")

# less than
print(f"less than operation: {i < j}")

# equal to
print(f"equal to operation: {i == j}")

# not equal to
print(f"Not Equal to operation: {i != j}")
````

### Inplace operations

````python
m = torch.rand(2, 3)
n = torch.rand(2, 3)

print(m)
print(n)
````

````python
m + n # this is being stored in a new tensor
````

````python
m.add_(n) # this will store m + n into m's place
````

````python
m.relu_() # this will store m's value after applying relu in m again
````

The trailing underscore after the functions add, relu will make it an inplace operation

### Copying a tensor

Assignment operator `=` makes copies the reference, not the value. Let's say, a is a tensor, we made another tensor b = a. Now if we make any changes into a, the same change will be applied to b as well. 
For avoiding this, we have to use `clone()` method. if we do
```bash
b = a.clone()
```
That will only copy the values of a into b, not the reference

````python
# permute
b = torch.rand(2, 3, 4)
print(b)
print(f"Shape of b: {b.shape}")
````

````python
# permute(2, 0, 1) means, in the main tenor = 2 was the 0th, 3 was the 1st and 4 was the 2nd dimension. We are calling
# permute(2, 0, 1) means, make the 2nd dimension as 0th, make the 0th as 1st, and 1 as 2nd
print(b.permute(2, 0, 1))
print(b.permute(2, 0, 1).shape)
````

````python
# unsqueeze - adds a new dimension in a given position
c = torch.rand(226, 226, 3) # image size
print(c)
print(c.unsqueeze(0).shape)
print(c.unsqueeze(1).shape)
print(c.unsqueeze(3).shape)
````
