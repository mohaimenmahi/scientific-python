# Pytorch Autograd
Autograd is a core component of PyTorch that provides automatic differentiation for tensor operations. It enables gradient computation, which is essential for training machine learning models using optimzation algorithms like gradient descent. 

### Why is Autograd that important?
It helps up with the complex derivatives we need in our machine learning work, avoids writting long derivative functions manually by hand.
A Neural Network is a collection of nested function, where we need to apply chain derivatives. Autograd comes here to help us out.


```python
import torch
```

```python
x = torch.tensor(3.0, requires_grad=True) 
# requires_grad=True makes the pytorch understand that we are going to calculate the derivative of the tensor
# it will keep track each of the states, so that it can calculate the gradient easily without repeating and returning back to each ofn the states while calculating
# it creates a computational graph for it's convenience so that while backward propagation, it can do the operations very fast
```

```python
y = x**2
```

```python
print(x)
print(y)
```

```python
# find the derivative
y.backward()
```

```python
x.grad # gradient in respect of x (dy/dx)
```

### Using Autograd for finding Loss derivative

```python
# Binary Cross-Entropy Loss for scalar
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8 # to prevent log(0)
    prediction = torch.clamp(prediction, epsilon, 1-epsilon)
    return -(target * torch.log(prediction) + (1-target) * torch.log(1-prediction))
```

```python
x = torch.tensor(6.7) # input
y = torch.tensor(0.0) # target
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
# Forward Pass
z = w * x + b 
z
```

```python
y_pred = torch.sigmoid(z)
y_pred
```

```python
loss = binary_cross_entropy_loss(y_pred, y)
loss
```

```python
loss.backward()
```

```python
print(w.grad) # loss gradient respect to w
print(b.grad) # loss gradient respect to b
```
