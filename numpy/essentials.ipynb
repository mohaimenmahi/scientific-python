### What is NumPy?

NumPy is a Python library that deals with arrays in computing language and it's very fast and very efficient.


```python
import numpy as np
import matplotlib.pyplot as plt
```

### Creating arrays


```python
a1 = np.array([3, 5, 7, 9]) # creating a plain array
a2 = np.zeros(10) # creates an array with 10 0s
a3 = np.ones(10) # creates an array with 10 1s
a4 = np.random.random(10) # creates random numbers between 0 and 1
a5 = np.random.randn(10) # creates arrays with 10 number that are distributed in Gaussian Distribution, mean 0, sd 1
a6 = np.linspace(0, 10, 100) # creates array of 100 numbers between 0 to 10, equally spaced
a7 = np.arange(0, 10, 0.02) # same as linspace, instead of providing the number of elements, it is provided the range between the numbers
```

```python
print(a1)
print(a2)
print(a3)
print(a4)
print(a5)
print(a6)
print(a7)
```

### Array Indexing/Slicing


```python
a1 = np.array([2, 4, 6, 8, 10])
a1[2]
```

```python
a1[2:] # takes every element from index 2 to the end
```

```python
a1[:-2] # starts from the beginning and takes till last 2
```

```python
a1[1:-2] 
```

```python
a1[a1>3] # takes only the elements those are greater than 3
```

```python
names = np.array(['Jim', 'Luke', 'Josh', 'Pete'])
names = np.append(names, 'P-Solver')
names
```

```python
first_letter_j = np.vectorize(lambda s: s[0])(names)=='J'
# lambda s: s[0] => a lambda function that takes s and returns s[0]
# np.vectorize is complicated. It acts like a for loop
# what it does above, it takes the names array, it creates a for loop
# and for each element from the names array, it creates another array
# with the first letter of each element
first_letter_j 
```

```python
names[first_letter_j]
```

```python
a1[a1 % 4 == 0]
```

### Calculus / Statistical Functions


```python
a1 = 2 * np.random.randn(10000) + 10
```

Mean, Standard deviation, and percentiles of arrays


```python
print(f"Mean of a1: {np.mean(a1)}")
print(f"Standard Deviation: {np.std(a1)}")
print(f"80th Percentile element: {np.percentile(a1, 80)}")
```

Integrals and Derivatives


```python
x = np.linspace(1, 10, 100)
y = 1 / x**2 * np.sin(x)
dydx = np.gradient(y, x)
y_intg = np.cumsum(y) * (x[1]-x[0]) # x[1] - x[0] = dx
```

```python
plt.plot(x, y)
plt.plot(x, dydx)
plt.plot(x, y_intg)
```
