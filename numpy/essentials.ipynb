### What is NumPy?

NumPy is a Python library that deals with arrays in computing language and it's very fast and very efficient.


```python
import numpy as np
import matplotlib.pyplot as plt
```

### Creating arrays


```python
a1 = np.array([3, 5, 7, 9]) # creating a plain array
a2 = np.zeros(10) # creates an array with 10 0s
a3 = np.ones(10) # creates an array with 10 1s
a4 = np.random.random(10) # creates random numbers between 0 and 1
a5 = np.random.randn(10) # creates arrays with 10 number that are distributed in Gaussian Distribution, mean 0, sd 1
a6 = np.linspace(0, 10, 100) # creates array of 100 numbers between 0 to 10, equally spaced
a7 = np.arange(0, 10, 0.02) # same as linspace, instead of providing the number of elements, it is provided the range between the numbers
```

```python
print(a1)
print(a2)
print(a3)
print(a4)
print(a5)
print(a6)
print(a7)
```

```python
names = np.array(['Jim', 'Luke', 'Josh', 'Pete'])
names = np.append(names, 'P-Solver')
names
```

```python
first_letter_j = np.vectorize(lambda s: s[0])(names)=='J'
# lambda s: s[0] => a lambda function that takes s and returns s[0]
# np.vectorize is complicated. It acts like a for loop
# what it does above, it takes the names array, it creates a for loop
# and for each element from the names array, it creates another array
# with the first letter of each element
first_letter_j 
```

```python
names[first_letter_j]
```

### Calculus / Statistical Functions


```python
a1 = 2 * np.random.randn(10000) + 10
```

Mean, Standard deviation, and percentiles of arrays


```python
print(f"Mean of a1: {np.mean(a1)}")
print(f"Standard Deviation: {np.std(a1)}")
print(f"80th Percentile element: {np.percentile(a1, 80)}")
```

Integrals and Derivatives


```python
x = np.linspace(1, 10, 100)
y = 1 / x**2 * np.sin(x)
dydx = np.gradient(y, x)
y_intg = np.cumsum(y) * (x[1]-x[0]) # x[1] - x[0] = dx
```

```python
plt.plot(x, y)
plt.plot(x, dydx)
plt.plot(x, y_intg)
```

##### Examples

Question 1:
Let y = e^(-x/10) sin(x). Consider 10000 x intervals in the range [0, 10].

1. Plot the function y vs x in the range [0, 10]
2. Compute the mean and standard deviation of y for x values in [4, 7]
3. For x in the range [4, 7], find the value y_m such that 80% of y values are less than y_m
4. Plot dy/dx vs x
5. Find the locations (x co-ordinate) where dy/dx = 0


```python
# 1
x = np.linspace(0, 10, 10000+1)
y = np.exp(-x/10) * np.sin(x)
plt.plot(x, y)
```

```python
# 2
y_range = y[(x>=4)*(y<=7)]
mean = np.mean(y_range)
sd = np.std(y_range)
print(f"Mean: {mean}, Standard Deviation: {sd}")
```

```python
# 3
y_m = np.percentile(y_range, 80)
print(y_m)
```

```python
# 4
dydx = np.gradient(y, x)
plt.plot(x, dydx)
```

```python
# trick for getting the points where dy/dx is 0
# we need to find the sign of each two consecutive values of dy/dx
# if the multiplication gets plus sign, that means they are going 
# to same direction (both plus or both minus)
# if the product is negative, so we found a location where 
# the graph changed it's location, which means that went through the value 0

## dydx[1:] * dydx[:-1] = array of the products of the two consecutive dydx values
## x = array of the x coordinate values.
## we need to find for which x, dydx is 0
```

```python
bool_arr = dydx[1:] * dydx[:-1] < 0
# this boolean array indicates where the values are less than 0
x[1:][bool_arr]
## x[1:] => as bool_arr is 1 less than x's length
## also, x[1:] is an array, and from this array, we need the
## values that has dxdy products are less than 0
## so using the bool_arr as index, we can find the values of True positions
```
